```python
import numpy as np
def G(Y):
  return np.array ([Y[1],Y[2],-0.5*Y[0]*Y[2]])

def RK3(Y, h):

    K1 = G(Y)

    K2 = G(Y + h*K1/2)

    K3 = G(Y - h*K1 + 2*h*K2)

    Y_new = Y + h*(K1 + 4*K2 + K3)/6

    return Y_new
```

We are given the Blasius equation:
$$
[
\frac{dY}{d\eta}=G(Y)
]
$$
$$
[
f'''+\frac12ff''=0
]
$$
with boundary conditions
$$
[
f(0)=0,\qquad f'(0)=0
]
$$

$$
[
f'(\infty)=1
]
$$
The problem is that we do not know \(f''(0)\).

We call this unknown value \(s\):

$$
[
\boxed{f''(0)=s}
]
$$

Our task is to find the correct value of \(s\) such that \(f'(\infty)=1\).

This is called the shooting method. We guess \(s\), solve the IVP, check the final value of \(f'\), and correct our guess

Here:
- Y[0] represents \(f\)
- Y[1] represents \(f'\)
- Y[2] represents \(f''\)

We have the system:
$$[
Y'=G(Y)
]
$$
where
$$
[
Y=
\begin{bmatrix}
f\\f'\\f''
\end{bmatrix}
]
$$

and
$$
[
G(Y)=
\begin{bmatrix}
f'\\
f''\\
-\frac12ff''
\end{bmatrix}
]
$$
The slope is the derivative of each dependent variable with respect to \(\eta\).

Therefore,
$$
[
\boxed{
G(Y)=
\begin{bmatrix}
df/d\eta\\
df'/d\eta\\
df''/d\eta
\end{bmatrix}
}
]
$$
